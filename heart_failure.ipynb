```python
# Importa as bibliotecas
import pandas as pd # Dataframe
import numpy as np # Numeros

# Atribui o arquivo a uma variavel
my_file = "heart.csv"

# Cria o dataframe
df = pd.read_csv(my_file)
df.head()
```

```python
# Verifica colunas no caso de resultados faltantes ou incompletos
for column in df.columns.values.tolist():
  total = df[column].isna().sum()
  if total > 1:
    print(f"{column} possui {total} resultados faltantes")
```

```python
# Sexo para bool
df['Sex'] = df['Sex'].map({'F': 0, 'M': 1})

# ChestPainType (One Hot Encoding)
dummy_variable = pd.get_dummies(df['ChestPainType'], prefix='ChestPain')
df = pd.concat([df, dummy_variable], axis=1)
df.drop('ChestPainType', axis=1, inplace=True)

# RestingECG (One Hot Encoding)
dummy_variable2 = pd.get_dummies(df['RestingECG'], prefix='RestingECG')
df = pd.concat([df, dummy_variable2], axis=1)
df.drop('RestingECG', axis=1, inplace=True)

# ExerciseAngina para bool
df['ExerciseAngina'] = df['ExerciseAngina'].map({'N': 0, 'Y': 1})

# ST_Slope (One Hot Encoding)
dummy_variable3 = pd.get_dummies(df['ST_Slope'], prefix='ST_Slope')
df = pd.concat([df, dummy_variable3], axis=1)
df.drop('ST_Slope', axis=1, inplace=True)

df.head()
```

```python
# Finaliza o tratamento dos dados e salva em um novo arquivo
path = "heart_encoded.csv"
df.to_csv(path, index=False)
```

```python
# Importa as bibliotecas
import pandas as pd # Dataframe
from sklearn.model_selection import train_test_split # Divisão dos dados
from sklearn.linear_model import LogisticRegression # Regressão Logística
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix # Métricas

df = pd.read_csv("heart_encoded.csv")
df.head()
```

```python
# Separar variáveis independentes (x) e dependente (y)
x = df.drop('HeartDisease', axis=1)
y = df['HeartDisease']

# Dividir os dados em conjuntos de treinamento e teste (80% treino, 20% teste)
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# Criar e treinar o modelo de Regressão Logística
model = LogisticRegression()
model.fit(x_train, y_train)
```

```python
# Fazer previsões no conjunto de teste
y_pred = model.predict(x_test)
print("\n--- Métricas de Avaliação ---")
print(f"Acurácia: {accuracy_score(y_test, y_pred):.2f}")
print("\n--- Relatório de Classificação ---")
print(classification_report(y_test, y_pred))
print("\n--- Matriz de Confusão ---")
print(confusion_matrix(y_test, y_pred))
```
